### Acessando elementos através do índice e fatiando `arrays numpy`.

```python
arr = np.arange(0, 30, 3)
arr
```

```python
arr[3] # Os valores do `numpy array` podem ser acessados através do indíce
```

```python
arr[2:5] # Traz os elementos no intervalo [2, 5[ ou seja, nos indíces 2, 3 e 4 (note o 5 não incluso)
```

```python
arr[:5] # Nada antes dos `:` entende que começa no 0 o intervalo [0, 1, 2, 3, 4]
```

```python
arr[2:] # Nada após `:` entende que vai até o fim do array o final do intervalo
```

```python
arr[2:] = 100 # Podemos fazer a manipulação (no caso atribuição) através dos `arrays` recuperados.
arr
```

### Indíce e fatia (_slice_) em `arrays` bidimensionais

```python
arr = np.arange(50).reshape((5, 10)) # Uma matriz 5x10
arr
```

```python
arr.shape
```

```python
# Notações de dois colchetes ou um colchete, sendo a de um mais simples
arr[:3][:] # [:3] Fatia as linhas da matriz, e em seguida [:] traz todas as colunas
            # Como utiliza [:] seria o equivalente a: `arr[:3]`
```

```python
arr[:2, 3:9] # Para fazer o slice da segunda dimensão separa os `slices` por `,` conforme a documentação do NumPy [8]
                # Neste caso fatia as duas primeiras linhas (indíce 0 e 1) em seguida as colunas no intervalo de [3, 9[.
                # Havia pausado a aula para pesquisar como fazer mas em seguida na aula essa abordagem é tratada.
```

```python
# Fatia o array, fazendo uma referência. 
arr2 = arr[:3]
arr2
```

```python
arr2[:] = 100 # Essa atribuição afeta a matriz em `arr` pois arr2 é uma referência (aponta para) 
                # dos elementos do array.
arr2
```

```python
arr
```

```python
arr = np.arange(50).reshape((5, 10)) # Reinicializando o `arr`
arr
```

```python
arr2 = arr[:3].copy() # ao utilizar o `copy` os elementos armazenados em `arr2` agora não são mais uma
                        # referência dos elementos em `arr` e sim uma cópia que referência novos elementos
                        # e não mais aponta para outro `array`.
arr2[:] = 100
arr2
```

```python
arr
```

```python
arr[1:4, 5:] # Uma fatia de 3 linhas e 5 colunas
                # Primeira dimensão [1, 4[ (linhas)
                # Segunda dimensão [5, ...[ (colunas)
```

### Operações em `array numpy`

```python
arr2 = arr[:3]
arr2[:] = 100 # Para seguir com exemplo da aula

bol = arr > 50 # Retorna resultados da operação booleana para cada elemento do `array`.
bol
```

```python
arr[bol] # Podemos utilizar o resultado da operação lógica como indíce onde serão retornados apenas 
            # os elementos que são verdadeiros dentro do `array numpy`.
```
