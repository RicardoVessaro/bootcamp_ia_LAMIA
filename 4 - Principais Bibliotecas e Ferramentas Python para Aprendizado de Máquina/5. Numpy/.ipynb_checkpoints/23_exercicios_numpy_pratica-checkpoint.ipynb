# Exercícios NumPy  

Agora que aprendemos sobre NumPy, vamos testar seu conhecimento. Começaremos com algumas tarefas simples, para depois entrarmos nas perguntas mais complicadas.

#### Importe NumPy como np

```python
import numpy as np
```

#### Crie uma matriz de 10 zeros

```python
# Utilizar `zeros` do `np` passando 10
np.zeros(10) 
```

#### Crie uma matriz de 10 ones

```python
# Utilizar `zeros` do `np` passando 10
np.ones(10)
```

#### Crie uma matriz de 10 cincos

```python
# Criar uma matriz com uns e então multiplicá-la por 5 onde a operação é feita elemento por elemento.
np.ones(10) * 5
```

#### Crie um array de inteiros de 10 até 50

```python
# Utilizar `arange` que cria um arranjo (`array`) passando o início do 
# intervalo como 10 e o fim como 51, uma vez que o fim do intervalo 
# não é incluso.
np.arange(10, 51)
```

#### Crie um array dos numeros pares de 10 até 50

```python
# Criar o `array` utilizando o `arange` pulando de dois em dois através
# do terceiro parâmetro `step`. Assim começa com um número par (10) e 
# termina em par.
np.arange(10, 51, 2)
```

#### Criei uma matriz 3x3 com valores variando de 0 até 8

```python
# Criar uma vetor de 0 a 9 (não inclusivo), e utilizar o `reshape` tranformando em uma 
# matriz 3x3 com o parâmetro (3, 3).
np.arange(0, 9).reshape((3, 3))
```

#### Crie uma matriz identidade 3x3

```python
# Para criar a matriz identidade podemos utilizar `eye` informando o número de linhas e colunas 
# no caso 3. 
np.eye(3)
```

#### Use NumPy para gerar números aleatórios entre 0 e 1

```python
# Utilizar o sub módulo `random` do NumPy, nele há a função `rand` que recebe quantos
# número serão gerados (o shape separado por vírgula conforme a documentação e a 
# quantidade de elementos), no caso uma dimensão com um elemento.
np.random.rand(1)
```

#### Use Numpy para gerar um array de 25 números aleatórios tirados de uma distribuição normal.

```python
# Para a distribuição normal utiliza-se a função `randn` do NumPy que também 
# está no módulo `random`.
np.random.randn(25)
```

#### Crie a seguinte matriz:

```python
# A forma proposta na aula de solução difere, utilizou-se:
# np.arange(0, 101) / 100

# A seguinte matriz é uma matriz 10x10 onde os elementos estão entre 0 e 1 
# com intervalo de 0.01 entre cada elemento. Para isso cria-se um intervalo
# entre 0.01 e 1.01 (1 incluso) onde é incrementado em 0.01 cada elemento
# por fim altera o formato do arranjo para uma matriz 10x10 utilizando 
# `reshape`.
np.arange(0.01, 1.01, 0.01).reshape(10, 10)
```

#### Crie um array de tamanho 20 igualmente espaçado entre 0 e 1.

```python
# Para criar arranjos igualmente espaçados utiliza-se a função `linspace` do NumPy, onde:
# 0 é o início do intervalo e 1 o fim do intervalo (inclusivo) e 20 é a quantidade de 
# elementos igualmente espaçados.
np.linspace(0, 1, 20)
```

## Indexação Numpy e Seleção

Agora você receberá algumas matrizes e será solicitado a replicar as saídas resultantes da matriz:

```python
mat = np.arange(1,26).reshape(5,5)
mat
```

```python
# ESCREVE O CÓDIGO AQUI

# Fatiar as últimas 3 linhas, trazendo todas às colunas com exceção da primeira.
# Começar na 3 linha (indíce 2) e na segunda coluna (indíce 1).
mat[2:,1:]
```

```python
# ESCREVE O CÓDIGO AQUI 
# 20 é o elemento que se encontra na quarta linha (indíce 3) na quinta coluna (indíce 4)
mat[3, 4]
```

```python
# ESCREVE O CÓDIGO AQUI 
# Trazer apenas os elementos das 3 primeiras linhas (indíces 0, 1, 2) na segunda coluna (indíce 1)
# obs: se utilizar [:3, 1] na indexação irá retornar um vetor ao invés de matriz.
mat[:3, 1:2]
```

```python
# ESCREVE O CÓDIGO AQUI 
# Retornar a quinta linha (última) da matriz, para o último elemento podemos utilizar `-1` como indíce.
mat[-1]
```

```python
# ESCREVE O CÓDIGO AQUI 
# Retornar as duas últimas linhas da matriz, agora utilizando a fatia `-2:` como indíce.
mat[-2:]
```

### Agora faça o seguinte

#### Obter a soma de todos os valores no "mat"

```python
mat.sum() # `sum` traz a soma de todos os valores da matriz
```

#### Obter o desvio padrão dos valores em mat

```python
# Para trazer o desvio padrão utilizar o `std`.
mat.std()
```

#### Obter a soma de todas as colunas em mat

```python
mat.sum(0) # Conforme a documentação em [9] podemos somar os valores com base na `axis` (eixo) 
            # sendo 0 o eixo que opera a soma entre os elementos de cada coluna.
```
