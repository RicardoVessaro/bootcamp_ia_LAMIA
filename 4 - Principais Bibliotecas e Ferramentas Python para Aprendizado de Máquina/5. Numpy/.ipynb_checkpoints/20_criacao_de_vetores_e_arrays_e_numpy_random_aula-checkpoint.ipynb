Para utilizar o `numpy` foi adicionado um kernel com um ambiente virtual do conda conforme [7].

```python
import numpy as np # Uma convenção para `numpy` é chamar ela de `np`.
```

### Criando Arrays

```python
minha_lista = [1, 2, 3]
minha_lista
```

```python
np.array(minha_lista) # Cria um `numpy array` de uma dimensão.
```

```python
# Agora temos uma lista de listas que caracteriza duas dimensões, ou seja uma matriz.
minha_matriz = [ [1, 2, 3], [4, 5, 6], [7, 8, 9] ]
```

```python
np.array(minha_matriz) # Mostra uma matriz (note os dois colchetes)
```

### Métodos `numpy` para criação de séries e matrizes

```python
# shift + tab para observar documentação do método.
np.arange(0, 10) # Cria uma sequência de números, semelhante ao `range` do python. Note o último elemento não incluso.
```

```python
np.arange(0, 10, 2) # 2 é o `step`, neste caso de dois em dois.
```

```python
arr = np.zeros(3) # Cria uma matriz com elementos zerados
arr
```

```python
arr = np.zeros((5, 5)) # Cria uma matriz 5x5 com todos os elementos zerados.
arr
```

```python
np.ones((3, 3)) # Cria uma matriz 3x3 prenchida por números um.
```

```python
np.eye(4) # Cria uma matriz identidade 4x4
```

```python
np.linspace(0, 10, 3) # Cria uma série em que os itens dentro do intervalo são igualmente espaçados. 
                        # Os números são definidos pelo terceiro parametro, no caso três números dentro da série.
```

```python
# a sub biblioteca `random` do `numpy` cria séries ou matrizes aleatórias com regras diferentes.

np.random.rand(5) # Escolhe 5 números aleatórios entre 0 e 1 com igual probabilidade de ser selecionado (distribuição uniforme).
```

```python
np.random.rand(5) * 100 # Agora são números de 0 a 100 por conta do `* 100`.
```

```python
np.random.rand(5, 4) # Cria uma matriz 5x4 com valores aleatórios entre 0 e 1.
```

```python
np.random.rand(5, 4, 3) # Uma matriz de 3 dimensões.
```

```python
np.random.randn(4) # `randn` utiliza uma distribuição normal com média 0 e desvio padrão 1 (Gaussiana) 
                    # sendo está mais aplicada na prática, se ennquadra em uma séries de amostras que
                    # são mais relevantes que uma distribuição uniforme.
```

```python
np.random.randint(0, 100, 10) # Uma série de números inteiros dentro de um intervalo de uma distribuição aleatória.
                                # Números entre [0, 10 [ (10 não incluso)
```

```python
np.round(np.random.rand(5) * 100, 0) # Equivalente ao `randint` acima utilizando o `round` que recebe uma série
                                        # e arredonda, o segundo parâmetro é a quantidade de casas depois da
                                        # vírgula, no caso 0. 
```

### Métodos e atributos

```python
# Reshape

arr = np.random.rand(25)
arr # vetor de 25 elementos
```

```python
arr.reshape((5, 5)) # Neste caso cria instancia uma nova série, sendo uma matriz 5x5
                        # A váriave `arr` não é sobreescrita apenas se for reatribuída (utilizando `=`)
```

```python
arr = arr.reshape((5, 5)) # `arr` reatribuído
arr
```

```python
arr.shape # mostra o tamanho do array numpy
```

```python
arr.max() # retorna o maior valor dentro do array numpy
```

```python
arr.min() # retorna o menor valor dentro do array numpy
```

```python
arr.argmax() # retorna o índice na qual o maior elemento se encontra
```

```python
arr.argmin() # retorna o índice na qual o menor elemento se encontra
```
