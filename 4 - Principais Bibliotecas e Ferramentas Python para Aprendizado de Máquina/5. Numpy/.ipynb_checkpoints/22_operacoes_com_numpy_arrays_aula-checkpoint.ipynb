```python
import numpy as np
```

### Operações Aritméticas entre `arrays`.

```python
arr = np.arange(0, 16)
arr
```

```python
arr + arr # Operação de soma indíce por indíce, importante ambos terem o mesmo tamanho.
```

```python
arr - arr
```

```python
arr * arr
```

```python
arr / arr # Neste caso de divisão incorreta há um aviso (`RuntimeWarning`), neste caso a operação retorna `nan`
```

```python
1 / arr # Neste caso retorno um infinito definido por `inf`. 
```

```python
np.array([0, 1]) / 0 # Ambas as mensagens onde 0 / 0 resulta em `nan` e 1 / 0 resulta em `inf`.
```

```python
arr ** 2 # Também pode ser fazer exponenciação.
```

### Operações com escalares

```python
arr + 100 # Soma 100 a todos os elementos do `array`.  
```

```python
arr * 30 # Operações aplicadas a todos os elementos do array.
```

### Métodos do NumPy.

```python
np.sqrt(arr) # `sqrt` aplica a raiz quadrada para todos os elementos
```

```python
np.exp(arr) # `exp` aplica exponenciação a todos os elementos do `array`.
```

```python
np.mean(arr) # `mean` para saber a média do `array`
```

```python
np.std(arr) # `std` para saber o desvio padrão (standard deviation) do `array`
```

```python
np.sin(arr) # `sin` traz o seno de cada elemento.
```

```python
np.max(arr) # `max` traz o valor máximo do `array`.
```

```python
np.min(arr) # `min` traz o valor mínimo do `array`.
```

```python
arr.max() # pode também ser utilizado para retornar o valor máximo
```

```python
arr.min() # pode também ser utilizado para retornar o valor mínimo
```
