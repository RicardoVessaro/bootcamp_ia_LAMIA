```python
import pandas as pd 
df = pd.DataFrame({'col1': [1, 2, 3, 4], 'col2': [444, 555, 666, 444], 'col3': ['abc', 'def', 'ghi', 'xyz'] })
df.head() # `head` mostra as primeiras 5 linhas do `DataFrame`. Como o indíce não foi informado ele é composto
            # por números inteiros.
```

```python
df['col2'].unique() # `unique` faz a seleção de dados únicos de uma coluna específica. Traz os dados sem repetir.
                    # semelhante ao `unique` do `numpy` em seguida.
```

```python
import numpy as np
np.unique(df['col2']) # `unique` do `numpy`.
```

```python
len(df['col2'].unique()) # Podemos utilizar o `len` para saber a quantidade (tamanho) de 
                            # dados únicos mas o pandas tem um método próprio chamado `nunique`.
df['col2'].nunique()
```

```python
df['col2'].value_counts() # `value_counts` retorna a quantidade que cada valor aparece no `DataFrame`.
```

```python
df[ (df['col1'] > 2) & (df['col2'] == 444) ] # Dois filtros em colunas diferentes para seleção de dados no 
                                                # `DataFrame`, podendo utilizar operadores como `==`.
```

```python
def vezes2(x):
    return x * 2

df['col1'].apply(vezes2) # `Com o método `apply` podemos executar funções nas colunas do `DataFrame`
                            # No caso definimos a função `vezes2` e a informamos como parâmetro no `apply`. 
                            # O `apply` funciona como se fosse um `map`.
```

```python
df['col3'].apply(len) # Podemos utilizar `built in` functions no `apply`. O `apply` aplica uma função
                        # que retorna um valor e recebe um argumento como parâmetro.
```

```python
df['col1'].apply(lambda x: x*x) # Também podemos passar uma `lambda` ao utilizar o `apply`.
```

```python
# Para deletar colunas podemos utilizar a função `built in` `del`
del df['col2']
df
```

```python
df.columns # Para acessar as colunas existentes no `DataFrame`. 
```

```python
df.index # Para acessar o indíce do `DataFrame`. Esse indíce foi inicializado da forma padrão logo o panda utiliza
            # o `RangeIndex`.
```

```python
# Reiniciando o `DataFrame` conforme a aula.
df = pd.DataFrame({'col1': [1, 2, 3, 4], 'col2': [444, 555, 666, 444], 'col3': ['abc', 'def', 'ghi', 'xyz'] })
df.head()
```

```python
df.sort_values(by='col2', inplace=True) # `sort_values` ordena o `DataFrame`, que também disponibilia o 
                                        # parâmetro `inplace`.
df
```

```python
df.isnull() # `isnull` verifica quais valores do `DataFrame` são nulos.
```

```python
df.dropna() # `dropna` apaga os valores que são `nan` conforme aulas anteriores.
```

```python
df = pd.DataFrame({'col1': [1, 2, 3, np.nan],
                    'col2': [np.nan, 555, 666, 444],
                    'col3': ['abc', 'def', 'ghi', 'xyz']})
df.head()
```

```python
df['col1'].fillna(value=df['col1'].mean()) # Podemos popular os valores `nan` utilizando `fillna` conforme 
                                            # aulas anteriores. Também possui o parâmetro `inplace`.
```

#### Sobre o `pivot_table`

```python
data = {'A': ['foo', 'foo', 'foo', 'bar', 'bar', 'bar'],
    'B': ['one', 'one', 'two', 'two', 'one', 'one'], 
    'C': ['x', 'y', 'x', 'y', 'x', 'y'],
    'D': [1, 3, 2, 5, 4, 1]}
df = pd.DataFrame(data)
df
```

```python
# `pivot_table` funciona de forma semelhante a uma tabela dinâmica do excel. Tu consegue reajustar a visualização
# de seu `DataFrame` e adicionar indíces multi nível.
df.pivot_table(values='D', index=['A', 'B'], columns=['C']) # `pivot_table` possui três parâmetros necessários para seu uso: 
                            # `values`: Os valores de qual coluna quero visualizar (no caso a D).
                            # `index`: Quais colunas quero que funcione como índice. (no caso A e B).
                            # `columns`: Os valores de qual coluna será usado como coluna (no caso C).
```
