```python
import numpy as np
import pandas as pd # Para importar o pandas, por padrão como `pd`
```

```python
# Serie é o objeto base do Pandas (versão mais simples do dado em Pandas)

labels = ['a', 'b', 'c']
```

```python
minha_lista = [10, 20, 30]
arr = np.array([10, 20, 30]) # arranjo NumPy
d = {'a': 10, 'b': 20, 'c': 30} 
```

```python

pd.Series(data=minha_lista, index=labels) # pd.`Series` inicializa um objeto `Series` onde:
                                            # `data`: É o conjunto de dados utilizado para criar a `series`.
                                            # `index`: É o conjunto de dados utilizado como indíce em relação aos dados da `series`.
```

```python
series = pd.Series(data=minha_lista, index=labels)
series['b'] # Acesso através do indíce, muito semelhante a um dicionário
```

```python
pd.Series(minha_lista, labels) # Em ordem, primeiro parâmetro é o `data` e o segundo o `index`.
```

```python
pd.Series(arr, labels) # Podemos utilizar `numpy arrays`.
```

```python
# As `series` não são limitadas a tipos numéricos, podemos passar `strings`, e até métodos.
pd.Series([sum, print, len])
```

```python
# Indíce é a referência para acessar os dados da `Serie`.
ser1 = pd.Series([1, 2, 3, 4], index=['EUA', 'Alemanha', 'URSS', 'Japão'])
ser1
```

```python
ser2 = pd.Series([1, 2, 3, 4], index=['Alemanha', 'EUA', 'Itália', 'Japão'])
ser2
```

```python
# Podemos realizar operações aritméticas entre `series` com base no indíce.
# As operações são realizadas por indíce, e não necessariamente precisão estar na mesma ordem,
# no caso de indíces que existem em uma `serie` e não em outra a operação retorna `NaN`.
ser1 + ser2
```
