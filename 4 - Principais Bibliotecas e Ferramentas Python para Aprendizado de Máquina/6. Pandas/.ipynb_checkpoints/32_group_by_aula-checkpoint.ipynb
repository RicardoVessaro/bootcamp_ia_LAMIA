```python
import pandas as pd
# Cria um DataFrame
data = {'Empresa':['GOOG', 'GOOG', 'MSFT', 'MSFT', 'FB', 'FB'], # Dados utilizados para aula.
        'Nome':['Sam', 'Charlie', 'Amy', 'Vanessa', 'Carl', 'Sarah'],
        'Venda':[200, 120, 340, 123, 243, 350]}
```

```python
df = pd.DataFrame(data) # Criamos um `DataFrame` com os dados (dicionário) fornecidos
df
```

```python
group = df.groupby('Empresa') # Retorna um objeto `pandas.api.typing.DataFrameGroupBy`.
                                # Neste caso o `DataFrame` foi agrupado pela coluna `Empresa`
group
```

```python
group.sum() # Soma as demais colunas por `Empresa` utilizando o objeto DataFrameGroupBy instanciado anteriormente.
```

```python
group.mean(numeric_only=True) # Traz a média dos valores da coluna `Venda` agrupado por `Empresa`
                                # Para que não de erro foi utilizado o parâmetro `numeric_only` como `True`
                                # conforme a documentação do `mean` (shift + tab).
                                # O `mean` não foi aplicado para a coluna `Nome` pois ela é uma coluna de texto.
```

```python
group.describe() # O método `describe` traz uma série de informações do agrupamento realizado
```

```python
group.count() # `count` Conta a quantidade de valores em cada coluna do agrupamento.
```

```python
# Agora agrupando pelo nome
group = df.groupby('Nome')
group 
```

```python
group.sum() # Soma o total de vendas de cada `Nome` (ou vendedor) por conta do agrupamento ser feito 
            # através da coluna 'Nome'.
```

```python
group.sum().loc['Amy'] # Com o método `loc` é possível trazer a quantidade de vendas de um 'Nome' em específico.
```
