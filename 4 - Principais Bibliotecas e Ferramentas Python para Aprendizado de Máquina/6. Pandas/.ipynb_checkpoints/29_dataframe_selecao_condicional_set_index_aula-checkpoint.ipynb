```python
import pandas as pd
import numpy as np
```

```python
from numpy.random import randn # importa o `randn` diretamente.
np.random.seed(101)
```

```python
df = pd.DataFrame(randn(5, 4), index='A B C D E'.split(), columns='W X Y Z'.split()) # Recria o mesmo Dataframe que o
                                                                                        # da última aula.
df
```

### Operações no indíce do `DataFrame`

```python
df > 0 # Semelhante as operações em `numpy array`. # Retorna um `DataFrame` com o resultado da operação.
```

```python
bol = df > 0
df[bol] # No `numpy` retornava um `array` redimensionado aqui os que não atendem recebem `NaN`.=
```

```python
df[df['W'] > 0] # Neste caso ele excluiu a linha C pois ela o valor na coluna W é menor que zero. Diferente do que 
                    # fazer a operação sobre todo o `DataFrame` e utilizar como indíce.
```

```python
# Posso retornar os valores da coluna Y as quais o valor de W é maior que zero. Como se aplicasse um filtro ou uma
# busca no `DataFrame`.
df[df['W'] > 0]['Y']
```

```python
# Código equivalente a `df[df['W'] > 0]['Y']`
bol = df['W'] > 0
df2 = df[bol]
df2['Y']
```

```python
# A palavra chave `and` não consegue lidar com as `series` do `pandas` para fazer a comparação entre `series` 
# utilize o operador `&`.
df[ (df['W'] > 0) & (df['Y'] > 1)] # Filtra onde os valores na coluna W são maiores que 0 E os valores da coluna
                                    # Y são maiores que 1.
```

```python
df[ (df['W'] > 0) | (df['Y'] > 1)] # Caso queira fazer uma operação `or` entre `series` utiliza o operador `|`
```

### Resetar o indíce do `DataFrame`

```python
# Para redefinir o indíce utiliza o método `reset_index` 
df.reset_index() # Neste caso o indíce é 0, 1, 2... da mesma forma que é definido o indíce caso não seja informado
                    # o parâmetro `index` na construção do `DataFrame`.
                    # Note também que o indíce que existia anteriormente foi transformado em uma coluna.
```

```python
df
```

```python
df.reset_index(inplace=True) # Para que as alterações sejam realizadas no próprio `DataFrame` lembre-se de utilizar
                                # o parâmetro `inplace` como `True`
df
```

```python
col = 'RS RJ SP AM SC'.split() # Criando uma nova lista de valores que serão utilizados como indíce no `DataFrame`.
col
```

```python
df['Estado'] = col # Adiciona uma nova coluna no `DataFrame` lembrando que ela deve ter o mesmo tamanho que as
                    # linhas do `DataFrame`.
df
```

```python
df.set_index('Estado', inplace=True) # Com o método `set_index` ele define a coluna informada como novo indíce do `DataFrame`.
df
```
