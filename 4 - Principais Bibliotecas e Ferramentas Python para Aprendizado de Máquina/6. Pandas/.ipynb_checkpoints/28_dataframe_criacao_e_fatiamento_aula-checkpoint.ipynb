```python
import pandas as pd
import numpy as np
```

```python
# `DataFrame` é o objeto principal do Pandas, sendo um conjunto de `Series`.
```

```python
np.random.seed(101) # Define a `seed` dos números aleatórios, assim eles serão os mesmos para qualquer computador.
```

### Inicializando e manipulando `DataFrames`

```python
# Utiliza-se `pd.DataFrame` para criar um `DataFrame`. 
# O primeiro parâmetro é o `data` onde foi passado uma matriz `numpy` 5x4.
# O segundo parâmetro é o `index` onde foi passado uma lista com cinco letras (mesmo número de linhas da 
# matriz) de A a E.
# O terceiro parâmetro é o `columns` que indica as colunas do `DataFrame` onde foi passado uma lista de
# quatro letras (mesmo número de colunas da matriz) de W a Z
df = pd.DataFrame(np.random.randn(5,4), index='A B C D E'.split(), columns='W X Y Z'.split())
df
```

```python
df['W'] # Podemos acessar as colunas dos `DataFrames` utilizando `[]`.
```

```python
type(df['W']) # Colunas de `DataFrames` são `series`.
```

```python
type(df) # É um `DataFrame`
```

```python
df[['W', 'Z']] # Podemos recuperar uma lista de colunas ao passar uma lista no indíce
```

```python
df.W # Podemos utilizar diretamente o `.` para acessar as colunas (não confundir com métodos e atributos do 
        # `DataFrame`, ou colocar nomes iguais ou semelhantes.
```

```python
df['new'] = df['W'] + df['X'] # Desta forma estamos criando mais uma coluna no `DataFrame` chamada `new`.
                                # O Valor populado é a soma das colunas 'W' e 'X' indíce a indíce.
df
```

```python
df.drop('new', axis=1) # Por padrão o `drop` utiliza o eixo (`axis`) 0, que se refere aos indíces.
                # Por isso para remover colunas definimos o eixo como 1
```

```python
df # Mas na referência não foi removido
```

```python
# Podemos reatribuir o `Dataframe` utilizando:
# df = df.drop('new', axis=1)
# Ou utilizar o parâmetro `inplace` como `True`.
df.drop('new', axis=1, inplace=True)
df
```

### Acessando valores nos `DataFrames`

```python
# Podemos acessar utilizando o método `loc` do `DataFrame`
df.loc['A'] # para acessar a linha inteira, retorna uma `series`.
```

```python
df.loc['A', 'X'] # para acessar a coluna de determinada linha com base no nome dos indíces e das colunas.
```

```python
df.loc[['A', 'B'], ['X', 'Y', 'Z']] # Podemos retornar várias linhas e várias colunas ao informar listas 
                                    # nos indíces.
```

```python
# Podemos utilizar o `iloc` para acessar os dados utilizando notações do NumPy.
df.iloc[1:4, 2:] # Utiliza indíces numéricos para localizar os valores.
```
