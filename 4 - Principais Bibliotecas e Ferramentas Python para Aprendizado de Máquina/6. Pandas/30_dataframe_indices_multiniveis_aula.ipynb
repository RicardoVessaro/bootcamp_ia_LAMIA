```python
import numpy as np
import pandas as pd
```

```python
# Níveis de índice
outside = ['G1', 'G1', 'G1', 'G2', 'G2', 'G2']
inside = [1, 2, 3, 1, 2, 3]
hier_index = list(zip(outside, inside)) # o `zip` dentro `list` cria uma lista de `tuple` onde ele associa cada valor
                                        # das listas por indíce.
hier_index
```

```python
hier_index = pd.MultiIndex.from_tuples(hier_index) # Desta forma é criado um objeto `MultiIndex` que representa um
                                                    # indíce multi nível.
hier_index
```

```python
df = pd.DataFrame(np.random.randn(6, 2), index=hier_index, columns=['A', 'B']) # Instancia um `DataFrame` com o multi indíce criado 
                                                            # anteriormente. Note que há o índice G1 e G2 e dentro
                                                            # dele o índice 1, 2 e 3
df
```

```python
# Para acessar os valores do `DataFrame` com `MultiIndex` utiliza-se o `loc` com anotações dos colchetes.
df.loc['G1']
```

```python
df.loc['G1'].loc[1] # Agora dentro do grupo G1 utilizamos o `loc` novamente para acessar os valores internos.
```

```python
# Os índices possuem nomes, no caso do `DataFrame` `df` não há nomes para eles.
df.index.names # para acessar o nome dos índices.
```

```python
df.index.names = ['Grupo', 'Numero'] # Desta maneira nomeamos ambos os indíces atribuindo a lista no atributo
                                        # `names` dentro de `index`.
df
```

```python
# `xs` vem de `cross section`
df.xs('G1') # Acessamos o índice G1 através do `xs`
```

```python
# A diferença do `xs` para o `loc` é que podemos acessar índices internos do `MultiIndex` como por exemplo: 1
df.xs(1, level='Numero') # Acessas o nível `'Numero'` sem precisar acessar o indíce acima que é o `'Grupo'`
                            # Para isso devemos informar o parâmetro `level` com o nome do indíce que queremos 
                            # acessar.
                            # Note que o `DataFrame` possui os indíce externo (`'Grupos'`) mas não o interno
                            # (`'Numero'`) pois ele já foi filtrado pelo `cross section` (`xs`).
```
