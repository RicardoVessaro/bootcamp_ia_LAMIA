** Difícil: quantas pessoas tem um cartão de crédito que expira em 2025? **

```python
# Para saber se um cartão de crédito expira em 2025 criamos uma função.
def expires_in_2025(cc_exp_date):
    # Verificamos se o ano de expiração é igual a 25. 
    # Para buscar o ano de expiração utilizamos o índice `3:` na `str`
    # que retorna os últimos dois digitos que representam o ano de 
    # expiração do cartão.
    return cc_exp_date[3:] == '25' # Retorna um boolean indicando se expira ou não.

# Testamos a função
assert expires_in_2025('10/25')
assert not expires_in_2025('02/20')

# Utilizamos o `sum` para somar a `série` booleana retornada do método `apply` que recebe 
# a função `expires_in_2025` como parâmetro.
sum(ecom['CC Exp Date'].apply(expires_in_2025))
```

```python
# Solução da video aula utilizou `lambda`
sum(ecom['CC Exp Date'].apply(lambda x: x[3:]) == '25')
```

** Difícil: quais são os 5 principais provedores de e-mail / hosts mais populares (por exemplo, gmail.com, yahoo.com, etc ...) **

```python
# Criamos uma função para manipular a coluna `Email` e retornar apenas o provedor do email.
def email_provider(email):
    # Dividimos a `str` através do `@` que separa o nome e provedor do email, e em seguida
    # retornamos o segundo elemento que representa o provedor acessando-o através do índice 1.
    return email.split('@')[1]

# Testa a função email_provider.
assert email_provider('teste@email.com') == 'email.com'

# Para saber os 5 principais emails podemos aplicar a função `email_provider` utilizando o `apply` criando assim
# uma `series` de provedores e através disso utilizar o `value_counts` para contar e ordenar a ocorrência desses
# provedores e por fim utilizar o `head` para trazer apenas os cinco primeiros.
ecom['Email'].apply(email_provider).value_counts().head()
```

```python
# Solução apresentada na video aula utilizando `lambda` e `iloc`
ecom['Email'].apply(lambda x: x.split('@')[1]).value_counts().iloc[:5]
```
