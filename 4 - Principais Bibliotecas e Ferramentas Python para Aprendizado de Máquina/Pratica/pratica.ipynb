```python
import pandas as pd

# Extrai os dados do arquivo `csv` para inicializar o `DataFrame`.
# O arquivo utulizado é separado por `;` por isso configurado a incialização com `sep=';'`.
# Os valores decimais são separados por `,` por isso configurado a inicializaçãom com `decimal=','`
# Utilizado a coluna `TICKER` como index do `DataFrame` através da configuração `index_col='TICKER'`,
# a coluna `TICKER` representa o código do ativo. 
acoes = pd.read_csv('statusinvest-busca-avancada-28-06-2026.csv', sep=';', decimal=',', index_col='TICKER') 

# Visualização rápida do `DataFrame` utilizando o método `head`.
acoes.head() 
```

```python
acoes.info() # Informações da estrutura do `DataFrame` após inicialização.
```

```python
# Função que formata os valores para float. Utilizada para converter diversas colunas do `DataFrame`
def str_to_float(str_value):
    # Como os valores vem com '.' para separar as unidades de milhar em diante e `,` para separar os centavos
    # essa `str` é tratada antes de ser transformada em `float`.
    str_value = str_value.replace('.', '')
    str_value = str_value.replace(',', '.')
    return float(str_value)

# Testando a função `str_to_float` com as possibilidades dentro do `DataFrame`
assert str_to_float('425.082.954,78') == 425082954.78
assert str_to_float('42,95') == 42.95
assert str_to_float('82') == 82
assert str_to_float('0.0') == 0
```

```python
# Váriaveis para referencias o nome das colunas mais utilizadas
LIQUIDEZ_MEDIA_DIARIA = ' LIQUIDEZ MEDIA DIARIA'
LIQUIDEZ_MEDIA_DIARIA_FLOAT = 'LIQUIDEZ MEDIA DIARIA FLOAT'
VALOR_DE_MERCADO = ' VALOR DE MERCADO'
VALOR_DE_MERCADO_FLOAT = 'VALOR DE MERCADO FLOAT'
DY = 'DY'
P_L = 'P/L'
PRECO = 'PRECO'
VPA = ' VPA'
P_VP = 'P/VP'
acoes.columns
```

```python
# Como pode ser necessário a conversào de várias colunas foi criado uma função para isso.
def column_to_float(dataframe, column, new_column=None):
    """ 
    Recebe um `DataFrame` como primeiro parâmetro e em seguida o 
    nome da coluna que será convertida (`column`). 

    Caso informado o parâmetro `new_column` uma nova coluna é 
    adicionada ao `dataframe` se não a coluna convertida é sobreescrita.
        
    """
    if new_column is None :
        new_column = column

    # Poderia criar uma lógica que traz um novo `DataFrame` sem sobreescrever 
    # o informado, semelhante a regra do parâmetro `inplace` mas não é o caso.
    dataframe[new_column] = dataframe[column].apply(str_to_float)

    return dataframe
```

```python
# Utilizaremos apenas registros que possui uma `LIQUIDEZ MEDIA DIARIA` que representa a média de liquidez do ativo
# nos últimos 30 dias [34]. 
acoes = acoes.dropna(subset=[LIQUIDEZ_MEDIA_DIARIA])

# Criado a coluna para representar a liquidez média diária em `float`.
# acoes[LIQUIDEZ_MEDIA_DIARIA_FLOAT] = acoes[LIQUIDEZ_MEDIA_DIARIA].apply(str_to_float)
column_to_float(acoes, LIQUIDEZ_MEDIA_DIARIA, LIQUIDEZ_MEDIA_DIARIA_FLOAT) # Manteve a coluna original para melhor visualização.

None # Apenas para não haver saida no jupyter e não poluir a visualização.
```

```python
# Removendo ações que não possuem valor de mercado.

# Primeiro removendo valores nulos do `DataFrame` para não considerar ações que não possuem valor de mercado.
acoes = acoes.dropna(subset=[VALOR_DE_MERCADO])

# Criado a coluna para representar o valor de mercado em `float`.
# acoes[VALOR_DE_MERCADO_FLOAT] = acoes[VALOR_DE_MERCADO].apply(str_to_float)
column_to_float(acoes, VALOR_DE_MERCADO, VALOR_DE_MERCADO_FLOAT) # Manteve a coluna original para melhor visualização.

# Filtra por acoes que possuem valor de mercado maior que 0. 
acoes = acoes [acoes[VALOR_DE_MERCADO_FLOAT] > 0]
```

```python
# Convertendo a coluna P/L para float.
# Convertido pois será utilizado a comparação com float mais a frente.
# acoes[P_L] = acoes[P_L].apply(str_to_float)
column_to_float(acoes, P_L)

None
```

```python
# Convertendo a coluna VPA para float para utilizar nos cálculos.
column_to_float(acoes, VPA)

None
```

Agora para fazer uma análise desses ativos com base em algumas das regras definidas pelo investidor Graham em [35] com objetivo de mostrar análises que podem ser úteis através das possibilidades que a biblioteca Pandas propõem.
1. Valor de mercado acima de 2 bi.
2. PL não pode ser superior a 15
3. Preço atual não ser maior que 1,5 * VPA.
4. PL * P/VP nao pode ser maior que 22,5

As regras escolhidas tem como objetivo mostrar as análises que pode ser feitas com base nas informações que temos no `DataFrame` a partir de um conceito apresentado, existem outras regras a respeito da escolha inicial de ativos que também devem ser consideras mas não são contempladas nesta análise de exemplo.

```python
graham = acoes.copy() # Copiamos o `DataFrame` para manter o `acoes` na forma tratada para ser usado por outras
                        # análises.
```

```python
# 1. Valor de mercado acima de 2 bi.
# Para isso basta filtrar o valor de mercado
graham = acoes[ acoes[VALOR_DE_MERCADO_FLOAT] > 2000000000 ] 
print(f'Quantidade de Ativos: {len(graham)}') # Apenas para mostrar a quantidade de ações na análise.
```

```python
# 2. PL nao pode ser superior a 15
# Para isso filtramos o P/L para apenas valores menores ou iguais a 15.
graham = graham[ graham[P_L] <= 15 ]
print(f'Quantidade de Ativos: {len(graham)}')
```

```python
# 3. Preço atual não ser maior que 1.5 * VPA (P>1.5*VPA)
# Primeiro calculamos a equação de VPA * 1.5 em seguida comparamos se o preço é menor ou igual ao resultado.
graham = graham[ (graham[VPA] * 1.5) <= graham[PRECO] ]
print(f'Quantidade de Ativos: {len(graham)}')
```

```python
# 4. PL * P/VP nao pode ser maior que 22,5
# De forma semelhante a regra três realiza-se o cálculo e em seguida seu valor é comparado conforme.
# A fim de ordenar os ativos por P/L * P/VP será adicionado a coluna `'PL_X_PVP'` no `DataFrame` armazenando o
# resultado da operação.
graham['PL_X_PVP'] = graham[P_L] * graham[P_VP]

# Comparado o resultado.
graham = graham[ graham['PL_X_PVP'] <= 22.5 ] 

# Mostramos o resultado ordenado.
# Ao mostrar o resultado foi observado que há resultados negativos, esses resultados serão removidos. 
graham = graham[ graham['PL_X_PVP'] > 0 ]

print(f'Quantidade de Ativos: {len(graham)}')
graham.sort_values(by='PL_X_PVP', ascending=False).head(10)
```

Agora filtraremos os ativos com base em uma regra definida por Bazin em [36] com o mesmo objetivo do filtro
anterior. A regra é a seguinte:
1. O Preço da ação pode ser no máximo 16,67 vezes o valor do dividendo.

Esse segundo `DataFrame` (`bazin`) será utilizado em conjunto do `graham` para que seja feito uma interseção
dos ativos que estão contemplados em ambas as regras, a fim de mostrar outras funcionalidades ensinadas nas 
aulas.

```python
bazin = acoes.copy() # Copiamos o `DataFrame` para manter o `acoes` na forma tratada para ser usado por outras
                        # análises.
```

```python
# Primeiro pegamos o valor do dividendo de forma hipotética multiplicando o `DY` pelo `PRECO` e adicionamos na coluna
# Dividendo.
# Primeiro removemos as linhas que não possuem `DY`.
bazin.dropna(subset=DY, inplace=True)

# Poderia fazer essa operação no próprio `DataFrame` `graham` mas a o objetivo e juntar as informações de duas fontes de dados.
bazin['DIVIDENDO'] = (bazin[DY] / 100 ) * bazin [PRECO] # Divide por 100 antes da multiplicação pois `DY` é uma porcentagem.

# Agora definimos o preço máximo permitido segunda a regra multiplicando o `DIVIDENDO` por 16.67
bazin['PRECO_MAXIMO'] = bazin['DIVIDENDO'] * 16.67

# Por fim basta filtrar comparando onde o `PRECO` é menor ou igual o `PRECO_MAXIMO` calculado.
bazin = bazin[ bazin['PRECO'] <= bazin['PRECO_MAXIMO'] ]

print(f'Quantidade de Ativos: {len(bazin)}')

# Para visualzação vamos ordenar pelos preços que possuem a maior diferença em relação ao preço máximo.
# Para isso vamos criar uma nova coluna chamada `DIFF_PRECO_MAXIMO` onde "DIFF" signifca diferença.
bazin['DIFF_PRECO_MAXIMO'] = bazin['PRECO_MAXIMO'] - bazin['PRECO']

bazin.sort_values(by='DIFF_PRECO_MAXIMO', ascending=False).head(10)
```

```python
# Agora com dois conjuntos de regras observamos quais os ativos que atendem ambos requisitos.

# Para isso podemos fazer o `merge` `inner` entre ambos os `DataFrames` através do indíce (`TICKER`)
# Para não adicionar colunas repetidas no `DataFrame` selecionamos apenas as colunas adicionais do `DataFrame` `bazin`.
analise = graham.merge(bazin[ ['DIVIDENDO', 'PRECO_MAXIMO', 'DIFF_PRECO_MAXIMO'] ], on='TICKER')

# Informações do `DataFrame` final.
analise.info()
```
