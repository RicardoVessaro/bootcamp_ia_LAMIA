## Solução Probabilidade Condicional
Primeiro nós modfiicamos o código para ter uma probabilidade de compra fixa independente da idade, digamos 40%:

```python
from numpy import random
random.seed(0)

totals = {20:0, 30:0, 40:0, 50:0, 60:0, 70:0}
purchases = {20:0, 30:0, 40:0, 50:0, 60:0, 70:0}
totalPurchases = 0

for _ in range(100000):
    ageDecade = random.choice([20, 30, 40, 50, 60, 70])
    purchaseProbability = 0.4 # Alteração feita na solução do professor. 
    totals[ageDecade] += 1
    if (random.random() < purchaseProbability):
        totalPurchases += 1
        purchases[ageDecade] += 1
```

Em seguida vamos calcular P(E|F) para uma casa da idade, vamos utilizar 30 anos novamente:

```python
PEF = float(purchases[30]) / float(totals[30])
print("P(purchase | 30s): " + str(PEF))
```

Agora vamos calcular P(E)

```python
PE = float(totalPurchases) / 100000.0
print("P(Purchase): " + str(PE))
```

P(E|F) é bem próximo de P(E), então podemos dizer que E e F são variáveis igualmente independentes.
