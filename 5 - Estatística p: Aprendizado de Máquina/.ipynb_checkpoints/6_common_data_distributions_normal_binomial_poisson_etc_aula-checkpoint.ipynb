## Exemplo de Distrobuições de Dados

### Distribuição Uniforme (_Uniform Distribuition_)

```python
import numpy as np
import matplotlib.pyplot as plt

values = np.random.uniform(-10.0, 10.0, 100000) # `low`: -10.0; é o o valor mais baixo possível.
                                                # `high`: 10.0; é o valor mais alto possível.
                                                # `size`: 100000; é o tamanho da distribuição.
# Monta e mostra o gráfico
plt.hist(values, 50)
plt.show() # Note como a população é distribuida de forma uniforme nos conjuntos (barras do gráfico). Uniforme tem uma probabilidade
            # muito próxima de um valor pertencer a qualquer um dos intervalo. Por isso há essa linha plana ao observar o gráfico.
```

### Normal (_Normal_) / Gaussiana (_Gaussian_)
Visualize a função de densidade de probabilidade

```python
from scipy.stats import norm # importa o módulo norm do submódulo stats da biblioteca SciPy
import matplotlib.pyplot as plt 

x = np.arange(-3, 3, 0.001) # Cria um arranjo (numpy array) onde os valores vão de -3 a 3 onde o incremento é de 0.001
plt.plot(x, norm.pdf(x)) # Mostra o gráfico para cada valor de x, onde y é a probabilidade de densidade da distribuição 
                            # normalizada é feita através de `norm.pdf(x)`.
```
