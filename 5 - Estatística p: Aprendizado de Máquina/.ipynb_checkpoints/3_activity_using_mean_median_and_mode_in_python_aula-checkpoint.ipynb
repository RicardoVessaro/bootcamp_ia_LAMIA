## Média vs. Mediana

Vamos criar dados falsos de receita, centralizado por volta de 27.000 com uma distribuição normal e desvio padrão (_standard deviation_) de 15.000, com 10.000 elementos. (Discutiremos mais sobre esses termos depois, caso não esteja familiarizado com eles).

Então, compute a média - Deve ser perto dos 27.000:

```python
import numpy as np # Importa o numpy como foi visto no card 4. 

# `incomes` representa as receitas vindas.
# 27000 é o centro (centre). Cria o conjunto em volta desse valor.
# 15000 é o desvio padrão, quanto se espalha.
# 10000 a quantidade de elementos (tamanho).
incomes = np.random.normal(27000, 15000, 10000)

np.mean(incomes) # média, conforme card 3 também.
```

Podemos dividir a receita em 50 conjuntos e mostrar em um histograma.

```python
# Comando para permitir mostrar os grafícos do matplotlib na linha do notebook
%matplotlib inline 
import matplotlib.pyplot as plt # Para mostrar gráficos foi utilizada a biblioteca Matplotlib, 
                                # utilizando o módulo `pyplot` referenciada como `plt`
plt.hist(incomes, 50) # Cria o histograma com as receitas (`incomes`) divididas em 50 intervalos (50 barras no gráfico).
plt.show() # Mostra o histograma.
```

```python
# Como temos uma distribuição semelhante a mediana estará próxima da média
np.median(incomes) # retorna a média das receitas
```

Agora nós adicionaremos a receita de Jeff Bezos no conjunto. Uma receita desigual.

```python
incomes = np.append(incomes, [1000000000]) # `append` adiciona a lista informada ([1000000000]) a uma lista
                                            # já existente (`income`)
                                            # Dessa forma desequilibramos o conjunto de receitas.
```

A mediana não vai mudar muito mas a média vai:

```python
np.median(incomes) # Ainda a um resultado perto dos 27.000
```

```python
np.mean(incomes) # A nova receita mostrou um valor muito acima, esse é o efeito dos _outlyers_ que podem
                    # existir no conjunto.
```

## Moda

Agora vamos criar um outro conjunto de dados falsos para representar a idade de 500 pessoas.

```python
ages = np.random.randint(18, high=90, size=500) # Distribuição com a mesma frequência de ocorrer 
                                                 # com 500 elementos que possuem os valores entre 18 e 90.
ages
```

```python
from scipy import stats # Importa o módulo SciPy para calcular a moda.
stats.mode(ages) # A moda é calculada através do método `mode` do submódulo `stats`.
# Na saída:
# ModeResult(mode=np.int64(74), count=np.int64(14))
# A moda (`mode` é 74 e a quantidade de vezes que se repete é de 14 (`count`).
# A saída pode alterar pois os dados foram gerados aleatoriamente.
```
