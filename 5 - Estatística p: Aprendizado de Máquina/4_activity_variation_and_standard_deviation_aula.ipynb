### Standard Deviation (desvio padrão) e Variance (variância)

Ao analisar um histograma dos dados de frequência apresentado, se pode notar que o formato do histograma pode dizer sobre como está distribuído os dados do conjunto, essa distribuição traz algumas respostas quando análisamos o que cada eixo do histograma representa.

No caso temos a quantidade de aviões que chegam por minuto, sendo a maioria 4 chegadas por minuto mas podemos ter poucos dias que chegaram 1 por minuto e outros que chegaram 12 por minuto, indicado a distribuição dos dados dentro do conjunto, sendo os desvios. Podemos notar que e mais provável ter 4 chegadas por minuto.

A variância nos indica o quão espalhado estão os dados do conjunto.

**Variância:**

A variância representada por _sigma ao quadrado_ (σ²) é a média da diferença da média ao quadrado. 

Exemplo da aula:
- Conjunto: (1, 4, 5, 4, 8)
- Primeiro encontramos a média: (1+4+5+4+8)/5 = 4.4
- Agora as diferenças entre os elementos e a média: (-3.4, -0.4, 0.6, -0.4, 3.6)
- Elevamos as diferenças ao quadrado: (11.56, 0.16, 0.36, 0.16, 12.96)
- Por fim ecnontramos a média dessas diferenças: σ² = (11.56 + 0.16 + 0.36 + 0.16 + 12.96) / 5 = 5.04

Elevamos as diferenças ao quadrado para que os resultados negativos não equiparem os positivos transformando os em positivos de uma maneira que seu valor tenha significado dentro do conjunto, e ao elevar ao quadrado, os valores que mais diferenciam da média (maior desvio) tem o peso melhor distribuído ao calcularmos a média, reprensetando melhor a influência de cada elemento no conjunto para encontrar o a variância (σ²).

**Desvio Padrão:**

Para encontrar o desvio padrão basta tirar a raiz da variância:
- σ² = 5.05
- σ = √5.04 = 2.24

Com o desvio padrão podemos identificar quais valores são considerados _outlyers_, ou seja não representam o conjunto e são considerados devios. Geralmente utiliza um σ para indicar, no caso valores que diferem da média somando ou subtraindo. No caso 2.24, mas podemos nos referir a quantos σ (_sigmas_) determinado valor está fora do padrão, ou distante da média. No exemplo podendo ser 1σ (2.24), 

**População (_population_) vs. Amostra (_sample_)**

Amostra é quando está trabalhando com parte do conjunto e não todo ele que seria a população. 

No caso de amostras ao tirar a média das diferenças que estão elevadas ao quadrado não as dividimos or N mas sim N-1. 

No caso do exemplo de variância, caso fosse uma amostra e não toda a população, utilizaria:
- N = 5
- S² = (11.56 + 0.16 + 0.36 + 0.16 + 12.96) / (N-1)
- S² = (11.56 + 0.16 + 0.36 + 0.16 + 12.96) / 4

Para calcular a variância da amostra representada por S².

**Notação Matemática**

Variância da população:

$$\sigma^2 = \frac{\Sigma(X - \mu)^2}{N}$$

Σ Significa soma;
_X_ Indica cada elemento do conjunto;
_N_  A quantidade total de elementos (tamanho) da população;
μ Representa a média (_mean_) da população.


Variância da amostra:
$$S^2 = \frac{\Sigma(X-M)^2}{N-1}$$
_M_ Representa a média da amostra.
_N_ A quantidade total de elementos (tamanho) da amostra.

## Desvio Padrão e Variância

```python
# Traduzi o notebook utilizado na aula, comentários meus relacionados a aula serão feitos em comentários nas tuplas de Python como está
```

```python
import numpy as np
import matplotlib.pyplot as plt

incomes = np.random.normal(100.0, 20.0, 10000) # Centro: 100.0; desvio: 20.0; quantidade: 10.000

plt.hist(incomes, 50)
plt.show()
# Criar as receitas e mostra o gráfico conforme aula 3.
```

```python
incomes.std() # Para saber o desvio padrão (standard deviation, std) 
                # Valor perto de 20.0 que foi o especificado ao criar o numpy array
```

```python
incomes.var() # Para saber a variância (variance)
                # Perto de 400 que é 20 ao quadrado. Pois o desvio padrão é a raiz quadrada da variância.
```
