Abaixo o código que irá gerar dados aleatórios de um e-commerce. Apenas um arranjo com o total gasto por transação. Selecione o bloco de código e pressione "play" para executá-lo.

```python
import numpy as np
import matplotlib.pyplot as plt

incomes = np.random.normal(100.0, 20.0, 10000) # Gera as receitas centralizados por volta de 100.00, com desvio de até 20.00, 
                                                # gerando 10000 elementos.

plt.hist(incomes, 50) # Divide o conjunto em 50 intervalos.
plt.show() # mostra o histograma, desta vez não utilizou o comando `%matplotlib inline`
```

Agora encontre a méddia e mediana desses dados. No bloco de código abaixo execute o seu código e veja se o resultado faz sentido:

```python
np.mean(incomes) # Para mostrar a média das receitas
```

```python
np.median(incomes) # Para mostrar a mediana das receitas. Note que ambas estão muito parecidas por conta da distribuição.
```

```python
# Para curiosidade vou adicionar alguns valores absurdos para bagunçar a receita (`incomes`). 

new_incomes = np.random.normal(100000, 30, 5) # Criamos mais 5 receitas por volta de 10.000 de valor com variação de 30.
print('new_incomes: ', new_incomes)

incomes = np.append(incomes, new_incomes) # Adicionamos as novas receitas às existentes.

plt.hist(incomes, 50) # Criamos novamente o gráfico agora com as novas receitas
plt.show() # Mostra o gráfico. Há já uma diferença no gráfico.
```

```python
# Temos poucos receitas novas (perto do valor de 1200), mas agora quando mostramos a sua média:
np.mean(incomes)
```

```python
np.median(incomes) # Temos uma diferença de aproximadamente 50 de valor entre a média e a mediana o que pode 
                    # mudar a análise da loja. Note que a mediana representa melhor os valores do primeiro gráfico (sem os desvios)
                    # E agora a média e quase um dos valores mais altos que possuia de receita antes de desviar os desvios.
```
