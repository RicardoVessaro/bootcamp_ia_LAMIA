```python
# Probabilidade condicional mede qual qual a probabilidade de algo acontecer dado que determinada coisa aconteceu antes que depende desta 
# segunda coisa acontecer.
```

```python
# Probabilidade Condicional (Conditional Probability)

# Qual a probabilidade de B acontecer dado que A aconteceu? -> P (B | A)?
# É calculada pela probabilidade de A e B acontecerem -> P(A, B) 
# Dividido pela probabilidade de A acontecer -> P(A)
# Notação matemática para isso:
```

$$ P(B | A) = \frac{P(A,B)}{P(A)}$$

```python
# Exemplo dado na aula:
# Um professor deu a seus alunos duas provas. 60% dos estudantes passaram nas duas provas, mas a primeira prova era mais fácil logo 80%
# dos alunos passaram nessa prova, qual a chance do aluno que passou na primeira prova passar também na segunda prova?

# A probabilidade de um aluno passar nas duas provas é de 60%: Logo P(A,B) é 0.6 -> probabilidade de ambos acontecerem. 
# A probabilidade de um aluno passar na primeira prova é de 80%: Logo P(A) é 0.8 -> probabilidade de A acontecer. l
# Então a chance de B acontecer (passar na segunda prova), uma vez que A aconteceu (aluno passou na primeira prova, B depende disto) é:
# 0.6 / 0.8 = 0.75; Logo P(B|A) é 0.75;
# A probabilidade de um estudante passar na segunda prova (B) uma vez que ele passou na primeira (A) é de 75%.
```

$$ P(B | A) = \frac{P(A,B)}{P(A)} = \frac{0.6}{0.8} = 0.75$$

# Probabilidade Condicional Atividade & Exercício

Abaixo temos um código para criar dados falsos sobre quanto as pessoas compram dado um intervalo de idade.

O código gera 100,000 "pessoas" aleatórias e aleatoriamente atribui a idade delas tendo a idade na casa dos 20, 30, 40, 50, 60 ou 70. 

Então atribui uma probabilidade menor de pessoas mais novas comprarem coisas.

No fim, temos dois dicionários Python:

`totals` contém o número total de pessoas em cada casa de idade. `purchases` contém o número total de coisas compradas por pessoas em cada grupo de idade. O total de todas as compras está em `totalPurchases` e nós sabemos que o nnúmero total de pessoas é 100,000.

Vamos rodar o código e dar uma olhada:

```python
from numpy import random
random.seed(0) # A seed para que os dados aleatórios sejam iguais cada vez que executado.

totals = {20:0, 30:0, 40:0, 50:0, 60:0, 70:0} # dicionário com os totais, iniciados com 0. Assim como as compras (`purchases`).
purchases = {20:0, 30:0, 40:0, 50:0, 60:0, 70:0}
totalPurchases = 0

for _ in range(100000): #`_` pois o valor não importa, apenas que itere 100000 vezes para criar os dados das pessoas.
    ageDecade = random.choice([20, 30, 40, 50, 60, 70]) # Escolhe a casa da idade
    purchaseProbability = float(ageDecade) / 100.0 # Chance menor para quem tem a casa da idade menor (função da casa para probabilidade)
    totals[ageDecade] += 1
    if (random.random() < purchaseProbability): # Quanto maior a chance de comprar, maior a chance de executar o bloco `if` assim é 
                                                # definido a relação entre a idade e a quantidade comprada.
        totalPurchases += 1
        purchases[ageDecade] += 1
```

```python
totals # Resultado igual da aula por conta do seed. O mesmo para `purchases` # Todas distribuidas quase igualmente
```

```python
purchases # Note que quanto maior a casa da idade maior as compras (relação)
```

```python
totalPurchases # Total de compras geradas aleatoriamente conforme a probabilidade
```

Agora vamos usar a probabilidade condicional

Primeiro vamos calcular P(E|F), onde E é a compra, e F é "você está na casa dos 30 anos". A chance de alguém na casa dos 30 anos comprar algo a porcentagem de quantas pessoas dentro dessa idade compraram algo. 

```python
PEF = float(purchases[30]) / float(totals[30])
print('P(purchase | 30s): ' + str(PEF)) # quase 30 por cento de chance devido o peso que calculamos. A probabilidade de você comprar algo
                                        # dado que está na casa dos 30 anos é de 30%
```

P(F) é a probabilidade de você estar na casa dos 30 anos neste conjunto de dados. 

```python
PF = float(totals[30]) / 100000.0 # Divide a quantidade de pessoas na casa dos 30 pelo total de pessoas no conjunto.
print("P(30's): " + str(PF))
```

E P(E) é a probabilidade geral de comprar algo independente da sua idade:

```python
PE = float(totalPurchases) / 100000.0 # Divide o total de compras pelo total de pessoas. 
print("P(Purchase):" + str(PE)) # 45% de chance de comprar algo
```

Se E e F são independentes, então esperamos que P(E | F) ser o mesmo que P(E). Mas eles não são; P(E) é 0.45, e P(E|F) é 0.3. Então, isso nos diz que E e F são dependentes (o que nós sabiamos nesse exemplo).

```python
# P(E|F) é diferente de P(E) pois a idade influencia na compra, por isso a diferença. Para sabermos se um valor depende do outro
# podemos utilizar essas contas.
```

P(E,F) é diferente de P(E|F). P(E,F) é a probabilide de você estar **tanto** na casa dos 30 **quanto** comprar algo, em relação a população total - não apenas entre a população das pessoas na casa dos 30, **sem um depender do outro**:

```python
print("P(30's, Purchase)" + str(float(purchases[30]) / 100000.0)) # P(E,F) é a chance de você comprar algo estando
                                                                    # na casa dos 30 em relação a todas as pessoas
```

Agora vamos calcular o produto de P(E) e P(F), P(E)P(F):


```python
print("P(30's)P(Purchase)" + str(PE * PF)) # Produto disso é a chance de você 
                                            # estar na casa dos trinta e também comprar algo.
```

Algo que você pode ter aprendido em estatistíca é que P(E,F) = P(E)P(F), mas isso assume que E e F são independentes. O que descobrimos é que P(E,F) é aproximadamente 0.5, enquanto P(E)P(F) é aproximadamente 0.75. Então quando E e F são dependentes - e demos uma probabilidade condicional acontecendo - nós não podemos dizer que P(E,F) = P(E)P(F).

Nós podemos confirmar que P(E|F) = P(E,F)/P(F), que é a relação mostrada nos slides - e certamente, é:

```python
print((purchases[30] / 100000.0) / PF)
```

# Sua Tarefa

Modifique o código acima onde a probabilidade da compra NÃO varie de acordo com a idade, fazendo com que E e F realmente sejam independentes. 

Então, confirme que P(E|F) é o mesmo que P(E), mostrando que a probabilidade condicional de comprar dado certa idade não é tão diferente quanto a probabilidade a-priori da compra apesar da idade.

```python
from numpy import random
random.seed(0)

totals = {20:0, 30:0, 40:0, 50:0, 60:0, 70:0}
purchases = {20:0, 30:0, 40:0, 50:0, 60:0, 70:0}
totalPurchases = 0

for _ in range(100000):
    ageDecade = random.choice([20, 30, 40, 50, 60, 70])
    # Mudança
    # -- remove
    # purchaseProbability = float(ageDecade) / 100.0 # Código anterior mudado para indicar onde ouve mudança,novo código é a linha abaixo)
    # ++ adiciona
    purchaseProbability = 0.4 # Definimos a chance de comprar algo em 40%, independente da idade.
    # fim da mudança
    totals[ageDecade] += 1
    if (random.random() < purchaseProbability):
        totalPurchases += 1
        purchases[ageDecade] += 1
```

```python
# Agora para saber se um independe do outro precisamos saber se P(E,F) é igual ao produto de P(E)P(F).

# P(E) = total de compras / total de pessoas
PE = totalPurchases / 100000
print(f'P(E) = P(compras): {PE}')

# P(F) = total de pessoas em uma idade / total de pessoas = A chance de ser uma pessoa em determinade idade 
PF = totals[30] / 100000
print(f'P(F) = P(estar em determinada idade): {PF}')

# P(E,F) Chance de comprar algo (E) com determinada idade (F), ou seja a compra feita por pessoas na casa dos 30 anos pelo total 
# de pessoas. 
PEF = purchases[30] / 100000
print(f'P(E,F): {PEF}')

# P(E)P(F)
PEPF = PE*PF
print(f'P(E)P(F): {PEPF}')

# Como os valores de P(E,F) e P(E) estão muito próximos é possível notar essa relação. 

# Agora vamos conferir P(E|F) é igual P(E)
PE_F = PEF / PF
print(f'P(E|F): {PE_F}')

# Note que P(E) é muito próximo a P(E|F) logo eles são indepentes um do outro.
```
