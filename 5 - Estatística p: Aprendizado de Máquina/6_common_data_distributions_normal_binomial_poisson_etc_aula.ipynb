## Exemplo de Distrobuições de Dados

### Distribuição Uniforme (_Uniform Distribuition_)

```python
import numpy as np
import matplotlib.pyplot as plt

values = np.random.uniform(-10.0, 10.0, 100000) # `low`: -10.0; é o o valor mais baixo possível.
                                                # `high`: 10.0; é o valor mais alto possível.
                                                # `size`: 100000; é o tamanho da distribuição.
# Monta e mostra o gráfico
plt.hist(values, 50)
plt.show() # Note como a população é distribuida de forma uniforme nos conjuntos (barras do gráfico). Uniforme tem uma probabilidade
            # muito próxima de um valor pertencer a qualquer um dos intervalo. Por isso há essa linha plana ao observar o gráfico.
```

### Normal (_Normal_) / Gaussiana (_Gaussian_)
Visualize a função de densidade de probabilidade

```python
from scipy.stats import norm # importa o módulo norm do submódulo stats da biblioteca SciPy
import matplotlib.pyplot as plt 

x = np.arange(-3, 3, 0.001) # Cria um arranjo (numpy array) onde os valores vão de -3 a 3 onde o incremento é de 0.001
plt.plot(x, norm.pdf(x)) # Mostra o gráfico para cada valor de x, onde y é a probabilidade de densidade da distribuição 
                            # normalizada é feita através de `norm.pdf(x)`.
                            # Note que no eixo x o 0 é a média e os demais números são os desvios padrões.
```

Gera números aleatórios com a distribuição normal. "mu" é a média desejada, "sigma" é o desvio padrão: 

```python
# importado novamente mas poderia ter importado tudo no início.
import numpy as np 
import matplotlib.pyplot as plt

mu = 5.0 # é a média (o 'centre'), por volta de 5.0 estará os valores dos elementos da população.
sigma = 2.0 # desvio padrão lembre-se do σ
values = np.random.normal(mu, sigma, 10000) # cria a população com a distribuição Gaussiana.
# Monta e mostra o gráfico.
plt.hist(values, 50)
plt.show() # Como são elementos aleatórios não será uma curva perfeita.
```

### Função de Densidade de Probabilidade Exponencial / "Lei de Potência"

```python
from scipy.stats import expon # importa o submódulo `expon` de `stats` da biblioteca SciPy. 
import matplotlib.pyplot as plt

x = np.arange(0, 10, 0.001) # Cria um arranjo que começa no 0 e vai até 10, com incremento de 0.001
plt.plot(x, expon.pdf(x)) # O gráfico indica uma queda exponencial. 
                            # Perto do 0 em x a chance de algo acontecer é mais alta (y), e essa chance vai diminuindo conforme 
                            # se distância de 0.
```

### Função de Massa de Probabilidade Binomial

```python
from scipy.stats import binom # O submódulo `stats` da biblioteca SciPY também tem um módulo para binomial (`binom`)
import matplotlib.pyplot as plt

n, p = 10, 0.5
x = np.arange(0, 10, 0.001) # Criamos valores discretos entre 0, 10 com incremento de 0.001
plt.plot(x, binom.pmf(x, n, p)) # `n`, e `p` são parâmetros para dar forma a função de massa de probabilidade
                                # `n` está para o eixo x (10) e p para o eixo `y`.
```

### Função de Massa de Probabilidade de Poisson
Exemplo: Meu site tem uma média de 500 visitas por dia. Qual a chance dele ter 550 visitas?

```python
from scipy.stats import poisson
import matplotlib.pyplot as plt

mu = 500 # `mu` é a média
x = np.arange(400, 600, 0.5) # Um intervalo onde a média é 500 [400, 600] com incremente de 0.5. 
plt.plot(x, poisson.pmf(x, mu)) # Utilziada para caso eu tenha uma média de um determinado período, eu consiga saber qual a probabilidade
                                # De outro valor em determinado período acontecer no futuro próximo.
```
