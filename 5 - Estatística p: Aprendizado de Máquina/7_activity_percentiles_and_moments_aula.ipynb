### Atividade
Experimente diferentes parâmetros ao criar dados de teste. Que feito eles tem nos percentils?

```python
def percentil_distribuicao_normal(mean, std, size=10000, percentile=50): 
    print(f'Média : {mean}, desvio padrão: {std}, tamanho: {10000}')
    vals = np.random.normal(mean, std, 10000) # Criamos conforem os parâmetros.
    print(f'Percentile {percentile}: {np.percentile(vals, percentile)}') # Vamos ver quanto é o valor do percentil. 
    plt.hist(vals, percentile)
    plt.show()
    return vals
```

```python
percentil_distribuicao_normal(30, 10, percentile=70)
```

```python
percentil_distribuicao_normal(30, 10, percentile=75) # Note que o percentil 75 ainda continua próximo da mediana
```

# Momentos: Media, Variância, _Skew_, e _Kurtosis_ 

```python
import numpy as np
import matplotlib.pyplot as plt

vals = np.random.normal(0, 0.5, 10000) # Uma distribuição normal para analisar os valores

# Mostrar o gráfico (50 intervalos)
plt.hist(vals, 50)
plt.show()
```

O primeiro momento é a média; Esta informação deve ser próximo de 0.

```python
np.mean(vals) # 1ro Momento (Média), perto de 0 pois é centralizado a 0 os dados.
```

O segundo momento é a variância.

```python
np.var(vals) # 2ndo Momento (Variância), O Desvio padrão é perto de 0.5 logo a variância é 0.25 
                # (raiz quadrada de 0.25 é 0.5)
```

O terceiro momento é o _skew_ - Como os dados são centralizados a 0, ele será próximo a zero.

```python
import scipy.stats as sp # Importa o SciPY como `sp` para calcular o Skew utilizando `skew`.
sp.skew(vals) # 3ro momento, o quão balanceado estão os dados.
```

O quarto momento é conhecido como "kurtosis", que define o formato da cauda. Para uma distribuição normal é 0.

```python
sp.kurtosis(vals) # `kurtosis` também do `scipy.stats`
                    # É próximo a 0 e é análisado comparando com a distribuição normal 
                    # (que é o tipo de distribuição em `vals`)
                    # Perto de 0 o topo é mais achatado com caudas mais alongadas (não tão finas).
```

## Atividade 
Entendendo _skew_: Mude a distribuição normal para ser centralizada próxima a 10 ao invés de 0 e veja que efeito isso tem nos momentos.
O _skew_ continua próximo a zero; _Skew_ é associado com o formato da distribuição e não o seu valor (_offset_) no eixo x.

```python
import numpy as np
import matplotlib.pyplot as plt
import scipy.stats as sp

def moments(mean, std, size=10000):
    gaussiana = np.random.normal(mean, std, size) # Uma distribuição normal (Gaussiana) por volta de 10 ao invés de 10. 
    plt.hist(gaussiana, 50)
    plt.show()
    
    print(f'(μ₁) Média:', gaussiana.mean() ) # 1ro momento (média)
    print(f'(μ₂) Variância (σ²):', gaussiana.var()) # 2ndo momento (variância)
    print(f'(μ₃) Skew (γ):', sp.skew(gaussiana)) # 3ro momento, continua próximo a zero conforme o explicado, pois não são os valores que 
                                                    # influenciam mas sim o formato que eles fazem (formato da distribuição)
                                                    # Está mais balanceada do que desequilibrada a distribuição por ser uma distribuição 
                                                    # normal.
    print(f'(μ₄) Kurtosis:', sp.kurtosis(gaussiana)) # 4rto momento, Também próximo a zero, logo não é um formato tão fino.
```

```python
moments(10, 0.5)
```

```python
moments(10, 2)
```

```python
moments(30, 0.5)
```

```python
moments(30, 2) # Se pode notar que o skew e o kurtosis não mudam conforme os valores, pois eles estão 
                # relacionado ao formato da distribuição (caulda e topo). 
                # Já o 1ro momento representa a média então ele muda conforme a média informada (centro) 
                # ao fazer uma distribuição Gaussiana aleatória, assim como o 2ndo momento é influeciado pelo 
                # desvio padrão, já que o segundo momento reprensenta a variância.
```
