```python
import numpy as np
import matplotlib.pyplot as plt
import scipy as scy
```

```python
plt.hist(np.random.uniform(-3, 3, 10000), 50) # Para colocar no relatório demonstrando a distribuição uniforme F1
plt.xlabel('valor')
plt.ylabel('quantidade')
plt.show()
```

```python
x = np.random.normal(0, 1, 10000) # Para colocar no relatório demonstrando a distribuição normal F2
plt.hist(x, 50)
plt.xlabel('valor')
plt.ylabel('quantidade')
plt.show()
```

```python
x = np.arange(0, 10, 0.001) # Para colocar no relatório demonstrando a lei de potência F3
plt.xlabel('eixo X')
plt.ylabel('eixo Y')
plt.plot(x, scy.stats.expon.pdf(x))
plt.show()
```

```python
x = np.arange(0, 10, 0.001) # Para colocar no relatório demonstrando a função de massa de probabilidade binomial F4
plt.xlabel('valor intervalo')
plt.ylabel('probabilidade')
plt.plot(x, scy.stats.binom.pmf(x, n=10, p=0.5))
plt.show()
```

```python
x = np.arange(100, 300, 0.5) # Para mostrar a Função de Massa de Probabilidade de Poisson no relatorio F5
mu = 200
plt.xlabel('número de pedidos p/ dia')
plt.ylabel('probabilidade')
plt.plot(x, scy.stats.poisson.pmf(x, mu))
plt.show()
```

```python
# Código para exemplo de gráfico de pizza para o relatório.
quantidade = [11, 26, 19, 7]
tamanhos = ['P', 'M', 'G', 'GG']
cores=['r', 'b', 'm', 'c']
plt.title('Quantidade de Camisetas por Tamanho')
plt.pie(x=quantidade, labels=tamanhos, colors=cores)
```

```python
plt.title('Quantidade de Camisetas por Tamanho')
plt.bar(tamanhos, quantidade, color=cores)
plt.show()
```

```python
x = np.random.randn(1000)
y = np.random.randn(1000)
plt.scatter(x, y) 
plt.show() # Mostrando para utilizar no relatório também
```

```python
x = np.random.rand(100) * 70 - 50 # Distribuição uniforme
outliers_positivos = np.random.rand(10) * 60 + 100 # adicionar alguns outliers para mostrar o boxplot
outliers_negativos = np.random.rand(10) * -60 - 100
data = np.concatenate((x, outliers_positivos, outliers_negativos))
plt.boxplot(data)
plt.show()
```
