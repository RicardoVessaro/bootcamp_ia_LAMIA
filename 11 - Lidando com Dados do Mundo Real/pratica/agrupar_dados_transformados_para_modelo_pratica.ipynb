# Atividade Prática 11 - Lidando com Dados do Mundo Real

Este _notebook_ é responsável por transformar os dados extraídos em um arquivo `.csv` final que 
será utilizado para programar o modelo.

```python
import pandas as pd

PATH_DIRETORIO = 'dados/bkp/final_08/'
```

A seguir foram lidos os arquivos `.csv` e mesclados (_merge_) utilizando 
Pandas.

```python
PATH_ARQUIVO_MUSICAS_PONTUACAO = PATH_DIRETORIO + 'kworb/transformacao/musicas_pontuacao_anual.csv'

musicas_pontuacao = pd.read_csv(PATH_ARQUIVO_MUSICAS_PONTUACAO, sep=';', decimal='.')

musicas_pontuacao.tail(3)
```

```python
PATH_ARQUIVO_MUSICAS_GENERO_0_1 = PATH_DIRETORIO + 'musicbrainz/transformacao/musicas_generos_0_1.csv'

musicas_generos_0_1 = pd.read_csv(PATH_ARQUIVO_MUSICAS_GENERO_0_1, sep=';', decimal=',')

musicas_generos_0_1.tail(3)
```

```python
dados_musicas = pd.merge(left=musicas_generos_0_1, right=musicas_pontuacao, on='id_musica')

dados_musicas.tail(3)
```

```python


colunas = list(dados_musicas.columns)

colunas_ordenadas = colunas[:2] + colunas[-4:] + colunas[2:-4]

colunas_ordenadas
```

```python
dados_musicas = dados_musicas[colunas_ordenadas]

PATH_ARQUIVO_DADOS_MUSICAS = PATH_DIRETORIO + 'dados_musicas.csv'
dados_musicas.to_csv(PATH_ARQUIVO_DADOS_MUSICAS, sep=';', decimal='.', index=False)

dados_musicas.tail()
```
