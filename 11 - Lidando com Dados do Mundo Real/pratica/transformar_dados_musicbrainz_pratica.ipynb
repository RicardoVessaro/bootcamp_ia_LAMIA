Este _notebook_ é responsável por transformar os dados extraídos de `extrair_dados_musicrainz_pratica`. 

Utiliando arquivo extraído vamos montar o arquivo que reuni os a música, gênero e pontuações para ser utilizado no
treinamento do modelo.

Músicas sem gênero não foram consideradas, apesar de remover os dados não ser uma boa prática o ideal seria utilizar um modelo que populasse os gêneros de acordo com outras caracterstícas, 
mas isso pode ser uma outra atividade por tanto foi utilizado apenas as músicas com genêro para realizar o teste do modelo.

```python
import pandas as pd

PATH_DIRETORIO = 'dados/final/musicbrainz/'

PATH_ARQUIVO_MUSICAS_GENEROS = PATH_DIRETORIO + 'musicas_generos.csv'

musicas_generos_df = pd.read_csv(PATH_ARQUIVO_MUSICAS_GENEROS, sep=';', decimal=',')

musicas_generos_df = musicas_generos_df.dropna().reset_index()[['id_musica', 'generos']].set_index('id_musica')

musicas_generos_df.tail(10)
```

Os genêros são separados por `|` foram criados colunas nos `DataFrames`
para cada gênero encontrado, dessa maneira a coluna irá possuir o valor de `0` indicando que 
não pertence a aquele gênero e `1` indicando que pertence.

```python
# 'id_musica' in musicas_generos_df.columns.values

with open(PATH_ARQUIVO_MUSICAS_GENEROS, 'r') as arquivo:
    first = True
    for linha in arquivo:
        if first : # Pula a primeira linha que é o cabeçalho
            first = False
            continue 
        
        id_musica, generos_str = linha.strip().split(';')

        generos = [] # Transforma os gêneros em lista para serem percorridos
        if generos_str != '':
            generos = generos_str.strip().split('|')

        #print(f'id_musica:{id_musica} , generos:{generos}')

        for genero in generos:

            if genero not in musicas_generos_df.columns.values: # Inicializa a coluna caso
                                                                # ainda não exista
                musicas_generos_df[genero] = 0

            musicas_generos_df.loc[id_musica, genero] = 1 # Atribui 1 nas colunas em que 
                                                            # a música pertence ao gênero.

# Warning por conta do tamanho do `DataFrame` pelo que entende-se na mensagem de aviso abaixo.
```
