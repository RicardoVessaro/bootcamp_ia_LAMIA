## Tratando a estrutura do ranque de músicas extraído

```python
import pandas as pd

ww_2026_df = pd.read_csv('dados/final/kworb/kworb_ww_2026.csv', sep=';', decimal='.')

ww_2026_df.head()
```

Nota-se que há registros repetidos, isso se dá pelo motivo de um artista participar da mesma música, mas a música ser diferente na tabela kworb por conta da participação dos artistas, para isso será considerado que o artista e a música são a mesma uma vez que ele está em ambas.

```python
ww_2026_df[ww_2026_df['musica'] == 'Usagi Rap'].head()
```

```python
usagi_rap = ww_2026_df[ww_2026_df['musica'] == 'Usagi Rap']
usagi_rap.iloc[0:5].groupby(by='musica')['pontos'].sum()
```

```python
usagi_rap.iloc[0:5]
```

Exemplos dos registros repetidos acontecem pois há músicas com quatro artistas onde dos quatro, três artistas se repetem, as linhas no kworb são as seguintes:

Usagi(CV:Ozawa Ari), Momonga(CV:Iguchi Yuka), Mermaid(CV:Oki Saeko) & Mermaid(CV:Nanase Ayaka) - Usagi Rap

Usagi(CV:Ozawa Ari), Mermaid(CV:Oki Saeko) & Mermaid(CV:Nanase Ayaka) - Usagi Rap

```python
repetidos = ww_2026_df[ww_2026_df['musica'] == 'Usagi Rap']

repetidos.head(20) # Há registros repetidos pois há músicas com 4 artistas onde o três artistas
```

```python
grupo_repetidos = repetidos.groupby(by=['artista', 'musica', 'data']).sum()
grupo_repetidos
```

Note que dessa maneira não há mais músicas repetidas para o artista, se um artista participa da mesma música em mais de uma vez é contando como uma música só.

```python
grupo_repetidos.xs('Usagi(CV:Ozawa Ari)', level='artista')
```

Agora será feito o tratamento para todo o `DataFrame`.

```python
ww_2026_df_group = ww_2026_df.groupby(by=['artista', 'musica', 'data']).sum()

ww_2026_df = ww_2026_df_group.reset_index() # Reset index faz com o que o `DataFrame` tenha o índice reiniciado 
                                            # para poder operar como `DataFrame`[1].
ww_2026_df.head(10)
```

## Separando os Artistas das Músicas mantendo o vínculo

### Criando novos `DataFrames` para vínculo

```python
import uuid

artistas = pd.DataFrame(ww_2026_df['artista'].unique(), columns=['artista']) # Criado para operacoes relacionadas a 
                                                                                # artista

artistas['id_artista'] = artistas['artista'].apply(lambda _: uuid.uuid7())

artistas.head(10)
```

```python
musicas = pd.DataFrame(ww_2026_df['musica'].unique(), columns=['musica']) # Fazer o mesmo para as músicas

musicas['id_musica'] = musicas['musica'].apply(lambda _: uuid.uuid7())

musicas.head(10)
```

```python
ww_2026 = pd.merge(left=ww_2026_df, right=artistas, on='artista')

ww_2026 = pd.merge(left=ww_2026, right=musicas, on='musica')

ww_2026
```

Para existir um vínculo entre artistas e e músicas foram selecionados as colunas do `DataFrame` com ambas as informações
e em seguida utilizado o método `drop_duplicates` do Pandas[2]. O índice foi reinicado apenas para recriá-lo.

### Salvando vínculo entre artistas e músicas

Salvando o vínculo

```python
musicas_artistas_df = ww_2026[['id_musica', 'musica', 'id_artista', 'artista']].drop_duplicates().reset_index()[['id_musica', 'musica', 'id_artista', 'artista']]

musicas_artistas_df.to_csv('dados/musicas_artistas.csv', sep=';', decimal='.', index=False) # Salvando o vínculo.

musicas_artistas_df.head()
```

```python
artistas.to_csv('dados/artistas.csv', sep=';', decimal='.', index=False) # Salvando artistas.

artistas.head()
```

```python
musicas.to_csv('dados/musicas.csv', sep=';', decimal='.', index=False) # Salvando artistas.

musicas.head()
```

Dados relacionadas a cada música salva, dessa maneira remove as músicas repetidas e mantém as informações diárias para
poder realizar os cálculos.

```python
musicas_pontuacao_dias = ww_2026[['id_musica', 'musica', 'data', 'pontos', 'pts_ganho', 'pts_calc']]

musicas_pontuacao_dias.to_csv('dados/musicas_pontuacao_dias.csv', sep=';', decimal='.', index=False) # Salvando músicas.

# musicas_pontuacao_dias.head()
musicas_pontuacao_dias[musicas_pontuacao_dias['musica'] == 'Usagi Rap'].head(30) # Para conferir se há repetição
```

## Transformando informações das pontuações.

```python
pontuacoes_df = pd.read_csv('dados/musicas_pontuacao_dias.csv', sep=';', decimal='.') # Lendo o `DataFrame` novamente 
                                                                                        # apenas para ter um ponto de 
                                                                                        # parada

pontuacoes_df
```

Agrupando o `DataFrame` por música pois as operações realizadas são por músicas.

```python
pontuacoes_group = pontuacoes_df[['id_musica', 'data', 'pontos', 'pts_ganho', 'pts_calc']].groupby(by=['id_musica'])

pontuacoes_group.head()
```

```python
musicas_df = pd.read_csv('dados/musicas.csv', sep=';', decimal='.', index_col='id_musica') # Indice é a música para 
                                                                                            # poder atribuir os valores
                                                                                            # resultantes do `groupby`

musicas_df.head()
```

```python
musicas_df['media_pontos'] = pontuacoes_group['pontos'].mean() # Média para utlizar a quantidade de pontos em média.
musicas_df['media_pts_ganho'] = pontuacoes_group['pts_ganho'].mean() # Média para utilizar a quantiade de pontos foram ganhos em média.
musicas_df['total_pts_calc'] = pontuacoes_group['pts_calc'].sum() # Acumulado uma vez que os `pts_calc` é dado por conta dos `pontos` e `pts_ganho`

musicas_df.head()
```

```python
musicas_df.to_csv('dados/musicas_pontuacao_anual.csv', sep=';', decimal='.')

musicas_df.head()
```
