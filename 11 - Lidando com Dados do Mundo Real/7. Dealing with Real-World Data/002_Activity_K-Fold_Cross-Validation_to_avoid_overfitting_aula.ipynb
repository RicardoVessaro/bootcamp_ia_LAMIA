# K-Fold Cross Validation

Vamos revisitar o _Iris data set_:


```python
import numpy as np
from sklearn import model_selection # Utilizado ao invés de `cross_validation` que está depreciado.
from sklearn import datasets
from sklearn import svm

iris = datasets.load_iris()

# REF:
# [1] https://scikit-learn.org/stable/api/sklearn.model_selection.html 
```

Uma simples divisão treino/teste é feita facilmente com a função `train_test_split` da biblioteca 
`model_selection`:

```python
# Divide os dados de iris em dados treino/teste com 40% reservado para teste
X_train, X_test, y_train, y_test = model_selection.train_test_split(iris.data, iris.target, test_size=0.4, random_state=0)
# X são os dados e Y são as classificações, divididos em dados para serem utilizados para treino e para teste.

# Monta um modelo SVC para predizer a classificação das iris utilizando os dados de treino.
clf = svm.SVC(kernel='linear', C=1).fit(X_train, y_train)

# Agora mede a performance do modelo com os dados de teste.
clf.score(X_test, y_test) # 96% de acerto no modelo.
```

_K-Fold cross validation_ é tão fácil quanto; vamos utilizar um _K_ de 5:

```python
# Nós damos a `cross_val_Score` um modelo, o conjunto de dados todo, os valores "reais", e o número de 
# dobras (K, conjuntos): 
scores = model_selection.cross_val_score(clf, iris.data, iris.target, cv=5) # `clf` é o modelo treinado anteriormente
                                                                            # `cv` indica a quantidade de conjuntos (K)
                                                                            # utilizados, onde um será utilizado para 
                                                                            # teste e os demais (K-1) para treino.

# Mostra a acurácia para cada conjunto (dobra):
print(scores)

# E a acurácia média das 5 (K) dobras (conjuntos):
print(scores.mean()) # Note que a acurácia foi de 96 para 98.
```

Nosso modelo é ainda melhor do que pensamos! Podemos fazer um modelo melhor? Vamos tentar um _kernel_ diferente
(_poly_):

```python
clf = svm.SVC(kernel='poly', C=1).fit(X_train, y_train) # Agora ao invés de `'linear'` esta sendo usado `'poly'`
                                                        # no parâmetro kernel.
scores = model_selection.cross_val_score(clf, iris.data, iris.target, cv=5)
print(scores)
print(scores.mean()) # Mostra 0.98 mas na aula mostrou 0.96 (acredito que seja por causa da versão da aula ser antiga)
```

No caso de o valor de acurácia fosse menor que 98 poderiamos dizer que o modelo piorou ao utilizar 
`kernel='poly'`, ou seja a opção polimonial estaria com _overfitting_. Mas no caso do simples treino/teste vemos a diferença:

```python
# Agora vamos medir a performance com os dados de teste
clf = svm.SVC(kernel='poly', C=1).fit(X_train, y_train)

# Agora medimos a performance com os dados de teste
clf.score(X_test, y_test) # Aqui vemos um _score_ de 90, abaixo dos 96 apresentado utilizando `linear`. Na aula foi o 
                            # mesmo resultado de 0.96 (novamente por conta da versão).
```

# Atividade

O "poly" _kernel_ para SVC possui outro atributo para o número de graus no polinomio utilizado, onde o padrão é 3. Por exemplo:

`svm.SVC(kernel='poly', degree=3, C=1)`.

```python
def poly_test(degree=3):
    '''
    Função para mostrar os resultados dos testes utilizando graus (degrees) diferentes.
    O modelo SVC é treinado e seu _score_ é validado utilizando `train/test` e `cross_val_score`.
    '''
    print(f'Poly: degree={degree}')
    p_clf = svm.SVC(kernel='poly', degree=degree, C=1).fit(X_train, y_train)
    p_scores = model_selection.cross_val_score(clf, iris.data, iris.target, cv=5)
    print('train/test score:')
    print(p_clf.score(X_test, y_test))
    print('cross_val_score:')
    print(p_scores)
    print(p_scores.mean()) 
    print('\n')
```

```python
poly_test(degree=1)
poly_test(degree=2)
poly_test()
poly_test(degree=4)
poly_test(degree=5)
poly_test(degree=6)
poly_test(degree=7)

# 2, 5, 6, 7 tiveram o mesmo resultado.
```
