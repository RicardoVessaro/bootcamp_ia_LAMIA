Páginas _web_ são compostas por HTML e CSS distribuidas pelo navegador quando acessamos algum endereço no navegador. Esses elementos podem ser acessados de forma programática acessando o fonte da página composto de HTML e CSS, isso é _web scraping_.

Nesta aula o _web scraping_ é feito com a biblioteca `BeautifulSoup4`.

```python
import requests # Utilizada para buscar o fonte da página web. Para istalar: pip install requests
from bs4 import BeautifulSoup as bs # Utilizado para web scraping. Para instalar: pip install beautifulsoup4
```

### Carregando nossa primeira página

```python
# Carrega o fonte da página web
r = requests.get("https://keithgalli.github.io/web-scraping/example.html")

# Converte o fonte para um objeto beautiful soup
soup = bs(r.content) # `soup` é uma instância da classe `BeautifulSoup4`.

# Print do HTML
# print(soup)
print(soup.prettify()) # Para mostrar o HTML identado. Mostra todoa a árvore de elementos da página. 
                        # A Documentação do Beautiful Soup está disponível em [4]
```

### `find` E `find_all`

```python
first_header = soup.find("h2") # Retorna o primeiro elemento H2 que encontrar no HTML da página.
print(first_header)

headers = soup.find_all("h2") # Retorna todos os elementos H2 que encontra no HTML da página. 
print(headers)

# "h2" é a consulta que será realizada outros elementos podem ser encontrados.
```

```python
# Passando uma lista de elementos para procurar
first_header = soup.find(["h2", "h1"]) # Lista de elementos para busca retornando o primeiro encontrado.
print(first_header)

headers = soup.find_all(["h2", "h1"]) # Lista retornando todos os elementos encontrados
print(headers)
```

```python
# Pode passar atributos para a busca utilizando `find`/`find_all`. 
paragraph = soup.find_all("p", attrs={"id": "paragraph-id"}) # Para isso utilizamos o parâmetro `attrs` que recebe um dicionário
                                                                # onde a chave é o atributo e o que deseja buscar fica como valor da propriedade.
print(paragraph)
```

```python
# Pode aninhar as buscas
body = soup.find("body")
print(body, '\n')

div = body.find("div")
print(div, '\n')

header = div.find("h1")
print(header, '\n')

# Desta forma a busca é feita afunilando entre o resultado encontrado, assim é navegado do elemento `body` para `div`; e 
# da `div` para `h1` que é o cabeçalho que está dentro da `div`.
```

```python
# Pode buscar por textos específicos, para isso se utiliza o parâmetro `string`.
paragraphs = soup.find_all("p", string="Some") # Buscou pelo texto exato de "Some" logo nada foi encontrado.
print(paragraphs) # Lista vazia pois nada foi encontrado (Retorna `None` quando é o metodo `find`).

# Para buscar textos que não seja exato utiliza-se expressões regulares.
import re # Pacote _built-in_ do Python para montar expressões regulares (RegEx, regex, regular expressions).

paragraphs = soup.find_all("p", string=re.compile("Some")) # `compile` constrói a regex. 
print(paragraphs)

headers = soup.find_all("h2", string=re.compile("(H|h)eader")) # Busca por cabeçalhos que possuem no seu texto a palavra `"header"`. Com a especificação
                                                                # `(H|h)` na _string_ o `compile` entende que buscamos por palavras Header e header
                                                                # H maiúsculo ou h minúsculo.
print(headers)
```

### `select` (Seletor CSS)

Parecido com o `find` e `find_all` mas a busca é feita utilizando uma busca semelhante a do CSS. Para entender as buscas em CSS temos a documentação em [3] e [4].

A busca é feito através dos seletores para aplicar um estilo em um elemento dentro do CSS, e o Beautiful Soup aplica a mesma busca desses seletores.

```python
print(soup.body.prettify()) # Para mostrar o `body` do HTML da página.
```

```python
content = soup.select("p") # De início pode achar muito parecido com o `find_all`.
print(content) 
print()

content = soup.select("div p") # Podemos pegar apenas elementos `p` que estão dentro de `div`, essa e a diferença de 
                                # usar o `select` ao invés de `find`
print(content) 
```

```python

paragraphs = soup.select("h2 ~ p") # Buscar todos os `p` que são precedidos pelo `h`, utilizando `~`.
print(paragraphs)
```

```python
# Podemos buscar elementos pelo identificador (id) utilizando `#`, util para pegar um elemento específico.
bold_text = soup.select("p#paragraph-id b") # Assim é pego o elemento _bold_ (`b`) que é precedido pelo elemento `p`
                                                # onde `p` possui o `id` com valor de `paragraph-id`.
bold_text
```

```python

paragraphs = soup.select("body > p") # Podemos buscar elementos onde um é descendente direto do outro utilizando `>`. 
                                        # Assim retorna elementos `p` que estão no primeiro nível do `body`.
print(paragraphs)
print()

# Em seguida podemos fazer uma estrutura aninhada iterando no laço do `for`.
for paragraph in paragraphs:
    print(paragraph.select("i"))
```

```python
# Podemos selecionar o elemento pela propriedade do CSS
soup.select("[align=middle]")
```

### Buscando diferentes propriedades do HTML

Ao invés de acessar o elemento inteiro, acessar propriedades dos elementos

```python
header = soup.find("h2")
print(header.string) # Podemos acessar o texto do elemento

div = soup.find("div")
print(div.prettify())
print(div.string) # Retorna `None` pois o elemento tem vários elementos que poder ser o texto de dentro dele
                    # para retornar os textos utilizar o `get_text`
print(div.get_text())
```

```python
# Acessar a propriedade específica do elemento.
link = soup.find("a")
link['href'] # Podemos acessar as propriedades através de colchetes (semelhante ao acesso do dicionário)
                # Assim com esse link podemos acessar outra página HTML e ir navegando entre diferentes páginas
                # utilizando o Beautiful Soup

# O mesmo pode ser feito para acessar o id do elemento
paragraphs = soup.select("p#paragraph-id")
print(paragraphs[0]['id'])
```

### Code Navigation (Navegação de Código)

Podemos utilizar os atributos HTML para navegar pela estrutura da página utiliando código. Métodos para navegar na 
árvore de elementos.

```python
print(soup.body.div.h1.string)
```

```python
print(soup.body.prettify())
```

Termos que precisamos saber se tratando de árvores HTML, estruturas aninhadas que as relações são descritas 
através dos níveis por:

**Parent**: Nivel acima do elemento em questão no caso acima o `body` é o _parent_ dos demais elementos. E a primeira
`div` é parent do `h1` que está dentro dele.

**Sibling**: Quando os elementos estão no mesmo nível, no caso os elementos `h2` e `h1` estão nos mesmos níveis, logo eles são _ siblings_. 

**Child**: Pegando o primeiro exemplo, também podemos dizer que a primeira `div` é _child_ do elemento `body` pois está no nível abaixo.

```python
soup.body.find("div").find_next_siblings() # `find_next_siblings` encontra os próximos _siblings_ da estrutura.
                                            # Na documentação há outros métodos para ajudar a navegar no HTML.
```
